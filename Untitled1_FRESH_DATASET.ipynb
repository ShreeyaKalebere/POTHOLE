### SECTION 2: Import Core Dependencies & Libraries

```python
import os
from pathlib import Path
import yaml
import pandas as pd
import numpy as np
import cv2
import matplotlib.pyplot as plt
from PIL import Image
from ultralytics import YOLO

print("Ultralytics YOLO and supporting libraries imported successfully!")
```

### SECTION 3: Dataset Configuration Verification (`data.yaml`)
Load and validate `F:\RoadDamage_Fresh\data.yaml` ensuring exactly 2 target classes: `pothole` (0) and `alligator_crack` (1).

```python
yaml_candidates = [
    Path("RoadDamage_Fresh/data.yaml"),
    Path("dataset/RoadDamage_Fresh/data.yaml"),
    Path("F:/RoadDamage_Fresh/data.yaml"),
    Path("/content/RoadDamage_Fresh/data.yaml")
]

yaml_path = None
for candidate in yaml_candidates:
    if candidate.exists():
        yaml_path = candidate
        break

assert yaml_path is not None, f"data.yaml not found in candidate paths: {yaml_candidates}"
print(f"Loading data.yaml from: {yaml_path.resolve()}")

with open(yaml_path, "r", encoding="utf-8") as f:
    yaml_config = yaml.safe_load(f)

print("\n--- data.yaml configuration ---")
print(yaml.dump(yaml_config, default_flow_style=False))

assert len(yaml_config.get("names", {})) == 2, "ERROR: Must define exactly 2 classes!"
assert yaml_config["names"][0] == "pothole", "ERROR: Class 0 must be pothole!"
assert yaml_config["names"][1] == "alligator_crack", "ERROR: Class 1 must be alligator_crack!"
print("data.yaml validated successfully: 0=pothole, 1=alligator_crack")
```

### SECTION 4: Dataset Integrity & Class Distribution Statistics
Verify directory layout and quantify bounding-box annotations across `train`, `val`, and `test` splits.

```python
dataset_root = Path(yaml_config["path"])
class_names = {0: "pothole", 1: "alligator_crack"}
stats = []

for split in ["train", "val", "test"]:
    img_dir = dataset_root / "images" / split
    lbl_dir = dataset_root / "labels" / split
    
    img_files = list(img_dir.glob("*.*"))
    lbl_files = list(lbl_dir.glob("*.txt"))
    
    pothole_boxes = 0
    alligator_boxes = 0
    negative_images = 0
    
    for lf in lbl_files:
        with open(lf, "r", encoding="utf-8") as f:
            lines = [l.strip() for l in f if l.strip()]
        if not lines:
            negative_images += 1
        else:
            for line in lines:
                parts = line.split()
                if parts:
                    cid = int(parts[0])
                    if cid == 0:
                        pothole_boxes += 1
                    elif cid == 1:
                        alligator_boxes += 1
                        
    stats.append({
        "Split": split,
        "Images": len(img_files),
        "Labels": len(lbl_files),
        "Pothole Boxes": pothole_boxes,
        "Alligator Crack Boxes": alligator_boxes,
        "Negative Images": negative_images,
        "Total Boxes": pothole_boxes + alligator_boxes
    })

stats_df = pd.DataFrame(stats)
print(stats_df.to_string(index=False))
```

### SECTION 5: Visual Inspection of Labeled Ground-Truth Samples
Render sample training images with ground-truth bounding boxes for both `pothole` and `alligator_crack`.

```python
import random
random.seed(42)

train_img_dir = dataset_root / "images" / "train"
train_lbl_dir = dataset_root / "labels" / "train"
labeled_imgs = [p for p in train_img_dir.glob("*.*") if (train_lbl_dir / f"{p.stem}.txt").stat().st_size > 0]
samples = random.sample(labeled_imgs, min(6, len(labeled_imgs)))

fig, axes = plt.subplots(2, 3, figsize=(16, 10))
axes = axes.flatten()

for idx, img_p in enumerate(samples):
    im = cv2.imread(str(img_p))
    im_rgb = cv2.cvtColor(im, cv2.COLOR_BGR2RGB)
    h, w, _ = im.shape
    
    lbl_p = train_lbl_dir / f"{img_p.stem}.txt"
    if lbl_p.exists():
        with open(lbl_p, "r") as f:
            for line in f:
                parts = line.strip().split()
                if len(parts) >= 5:
                    cid = int(parts[0])
                    xc, yc, bw, bh = map(float, parts[1:5])
                    x1 = int((xc - bw/2) * w)
                    y1 = int((yc - bh/2) * h)
                    x2 = int((xc + bw/2) * w)
                    y2 = int((yc + bh/2) * h)
                    color = (255, 140, 0) if cid == 0 else (0, 150, 255)
                    label = class_names.get(cid, str(cid))
                    cv2.rectangle(im_rgb, (x1, y1), (x2, y2), color, 3)
                    cv2.putText(im_rgb, label, (x1, max(20, y1-8)), cv2.FONT_HERSHEY_SIMPLEX, 0.7, color, 2)
                    
    axes[idx].imshow(im_rgb)
    axes[idx].set_title(img_p.name, fontsize=10)
    axes[idx].axis("off")

plt.tight_layout()
plt.show()
```

### SECTION 6: Initialize YOLO11 Pretrained Architecture
Load fresh pretrained `yolo11n.pt` backbone. DO NOT resume old training.

```python
model = YOLO("yolo11n.pt")
print(f"YOLO11 Model Architecture: {model.model.__class__.__name__}")
print("Pretrained YOLO11 Nano weights loaded successfully for fresh start!")
```

### SECTION 7: Fresh YOLO11 Training Execution
Train with `epochs=100`, `imgsz=640`, `batch=16`, `device=0`, `patience=20`, saving to `runs/detect/road_damage_fresh`.

```python
device_id = 0 if torch.cuda.is_available() else "cpu"
root_csv = Path("training_results.csv").resolve()

def on_epoch_end_cb(trainer):
    try:
        if hasattr(trainer, "csv") and trainer.csv.exists():
            import shutil
            shutil.copy2(str(trainer.csv), str(root_csv))
            print(f"\n>>> [EPOCH {trainer.epoch + 1}/{trainer.epochs}] Synced to {root_csv.name}")
    except Exception as e:
        pass

model.add_callback("on_fit_epoch_end", on_epoch_end_cb)

training_results = model.train(
    data=str(yaml_path.resolve()),
    epochs=100,
    imgsz=640,
    batch=16,
    device=device_id,
    project=str(Path("runs/detect").resolve()),
    name="road_damage_fresh",
    exist_ok=True,
    patience=50,
    save=True,
    save_period=10,
    plots=True,
    verbose=True
)

if (Path(training_results.save_dir) / "results.csv").exists():
    import shutil
    shutil.copy2(str(Path(training_results.save_dir) / "results.csv"), str(root_csv))

print("\n Training completed successfully!")
print(f"Results saved to directory: {training_results.save_dir}")
print(f"All epoch results available in CSV: {root_csv}")
```

### SECTION 8: Resolve Best Weights & Checkpoint Validation

```python
best_weight_path = Path(training_results.save_dir) / "weights" / "best.pt"
if not best_weight_path.exists():
    found_weights = sorted(list(Path("runs/detect/road_damage_fresh").glob("**/weights/best.pt")), key=lambda p: p.stat().st_mtime, reverse=True)
    if found_weights:
        best_weight_path = found_weights[0]

print(f"Resolved Best Model Weights: {best_weight_path.resolve()}")
assert best_weight_path.exists(), f"best.pt not found at {best_weight_path}"

best_model = YOLO(str(best_weight_path))
```

### SECTION 9: Comprehensive Validation Metrics & Per-Class Breakdown
Evaluate on validation split: Precision, Recall, mAP@50, mAP@50:95, and F1-score for each class.

```python
val_metrics = best_model.val(data=str(yaml_path.resolve()), split="val", device=device_id)

precision_overall = val_metrics.box.mp
recall_overall = val_metrics.box.mr
map50_overall = val_metrics.box.map50
map5095_overall = val_metrics.box.map
f1_overall = (2 * precision_overall * recall_overall) / (precision_overall + recall_overall + 1e-16)

print("\n================ VALIDATION METRICS OVERALL ================")
print(f"Precision: {precision_overall:.4f}")
print(f"Recall:    {recall_overall:.4f}")
print(f"F1-Score:  {f1_overall:.4f}")
print(f"mAP@50:    {map50_overall:.4f}")
print(f"mAP@50:95: {map5095_overall:.4f}")

class_map50 = val_metrics.box.maps50
class_map = val_metrics.box.maps
print("\n--- Per-Class Breakdown ---")
for cid, cname in class_names.items():
    c_m50 = class_map50[cid] if cid < len(class_map50) else float('nan')
    c_m = class_map[cid] if cid < len(class_map) else float('nan')
    print(f"Class {cid} ({cname}): mAP@50 = {c_m50:.4f}, mAP@50:95 = {c_m:.4f}")
```

### SECTION 10: Unseen Test Split Evaluation

```python
test_metrics = best_model.val(data=str(yaml_path.resolve()), split="test", device=device_id)

print("\n================ TEST SET METRICS ================")
print(f"Precision: {test_metrics.box.mp:.4f}")
print(f"Recall:    {test_metrics.box.mr:.4f}")
print(f"mAP@50:    {test_metrics.box.map50:.4f}")
print(f"mAP@50:95: {test_metrics.box.map:.4f}")
```

### SECTION 11: Display Evaluation Curves & Confusion Matrix

```python
run_dir = Path(training_results.save_dir)
plot_candidates = ["confusion_matrix.png", "PR_curve.png", "F1_curve.png", "results.png"]

for plot_name in plot_candidates:
    p = run_dir / plot_name
    if p.exists():
        plt.figure(figsize=(10, 8))
        im = cv2.imread(str(p))
        im_rgb = cv2.cvtColor(im, cv2.COLOR_BGR2RGB)
        plt.imshow(im_rgb)
        plt.title(plot_name, fontsize=14)
        plt.axis("off")
        plt.show()
```

### SECTION 12: Real-World Unseen Kolhapur / Local Evaluation
Evaluate the trained model on real-world unseen road conditions.

```python
real_world_dir = Path("real_world_test")
if real_world_dir.exists():
    rw_images = list(real_world_dir.glob("*.jpg"))
    print(f"Found {len(rw_images)} unseen real-world test images.")
    
    rw_results = best_model.predict(source=str(real_world_dir), conf=0.25, save=False)
    
    fig, axes = plt.subplots(len(rw_results[:4]), 1, figsize=(12, 6 * len(rw_results[:4])))
    if len(rw_results[:4]) == 1:
        axes = [axes]
    
    for idx, r in enumerate(rw_results[:4]):
        im_bgr = r.plot()
        im_rgb = cv2.cvtColor(im_bgr, cv2.COLOR_BGR2RGB)
        axes[idx].imshow(im_rgb)
        axes[idx].set_title(f"Real-World Inference: {Path(r.path).name} ({len(r.boxes)} detections)")
        axes[idx].axis("off")
        
    plt.tight_layout()
    plt.show()
else:
    print("No real_world_test directory found.")
```
